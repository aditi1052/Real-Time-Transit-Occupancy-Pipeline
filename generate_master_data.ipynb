```python
import pandas as pd
from datetime import datetime, timedelta
from pathlib import Path
```

```python
Output_Path = Path(
    "/Volumes/transit_occupancy/reference/source_files"
)

Output_Path.mkdir(parents=True, exist_ok=True)

print(f"Reference data will be written to: {Output_Path}")
```

```python
vehicle_capacities = [40, 50, 60]

vehicles = []

for vehicle_number in range(1, 21):
    capacity = vehicle_capacities[(vehicle_number - 1) % len(vehicle_capacities)]

    vehicles.append({
        "vehicle_id": f"BUS-{vehicle_number:03d}",
        "vehicle_type": (
            "Mini"
            if capacity == 40
            else "Standard"
            if capacity == 50
            else "Large"
        ),
        "vehicle_capacity": capacity,
        "registration_number": f"MH12-RT-{1000 + vehicle_number}",
        "service_status": "ACTIVE"
    })

vehicles_df = pd.DataFrame(vehicles)

vehicles_df.to_csv(
    Output_Path / "vehicles.csv",
    index=False
)

display(vehicles_df)
```

```python
stops_df = pd.DataFrame(
    stops,
    columns=[
        "stop_id",
        "stop_name",
        "latitude",
        "longitude",
        "zone"
    ]
)

stops_df.to_csv(
    Output_Path / "stops.csv",
    index=False
)

display(stops_df)
```

```python
route_stop_definitions = {
    "R-01": [
        "STOP-001", "STOP-002", "STOP-003",
        "STOP-004", "STOP-005"
    ],
    "R-02": [
        "STOP-006", "STOP-007", "STOP-008",
        "STOP-009", "STOP-010"
    ],
    "R-03": [
        "STOP-011", "STOP-012", "STOP-013",
        "STOP-014"
    ],
    "R-04": [
        "STOP-002", "STOP-003", "STOP-004",
        "STOP-005", "STOP-015"
    ],
    "R-05": [
        "STOP-016", "STOP-017", "STOP-015",
        "STOP-018", "STOP-014", "STOP-019",
        "STOP-020"
    ]
}

route_stops = []

for route_id, stop_ids in route_stop_definitions.items():
    for stop_sequence, stop_id in enumerate(stop_ids, start=1):
        route_stops.append({
            "route_id": route_id,
            "stop_id": stop_id,
            "stop_sequence": stop_sequence
        })

```

```python
route_stops_df = pd.DataFrame(route_stops)

route_stops_df.to_csv(
    Output_Path / "route_stops.csv",
    index=False
)

display(route_stops_df)
```

```python
trips = []

vehicle_number = 1
base_time = datetime.strptime("06:00", "%H:%M")

for route_number in range(1, 6):
    route_id = f"R-{route_number:02d}"

    for trip_number in range(1, 9):
        start_time = base_time + timedelta(minutes=90 * (trip_number - 1))
        end_time = start_time + timedelta(minutes=75)

        trips.append({
            "trip_id": f"TRIP-{route_number:02d}-{trip_number:02d}",
            "route_id": route_id,
            "vehicle_id": f"BUS-{vehicle_number:03d}",
            "scheduled_start_time": start_time.strftime("%H:%M:%S"),
            "scheduled_end_time": end_time.strftime("%H:%M:%S"),
            "direction": (
                "OUTBOUND"
                if trip_number % 2 != 0
                else "INBOUND"
            ),
            "trip_status": "SCHEDULED"
        })

        vehicle_number += 1

        if vehicle_number > 20:
            vehicle_number = 1

```

```python
trips_df = pd.DataFrame(trips)

trips_df.to_csv(
    Output_Path / "trips.csv",
    index=False
)

display(trips_df)
```

```python
datasets = {
    "vehicles.csv": vehicles_df,
    "routes.csv": routes_df,
    "stops.csv": stops_df,
    "route_stops.csv": route_stops_df,
    "trips.csv": trips_df
}

print("Generated reference datasets:")
print("-" * 45)

for file_name, dataframe in datasets.items():
    file_path = Output_Path / file_name

    print(
        f"{file_name:<20} "
        f"rows={len(dataframe):<5} "
        f"exists={file_path.exists()}"
    )
```
