```python
%sql

CREATE CATALOG IF NOT EXISTS transit_occupancy;

CREATE SCHEMA IF NOT EXISTS transit_occupancy.reference;             
CREATE SCHEMA IF NOT EXISTS transit_occupancy.bronze;
CREATE SCHEMA IF NOT EXISTS transit_occupancy.silver;
CREATE SCHEMA IF NOT EXISTS transit_occupancy.gold;
CREATE SCHEMA IF NOT EXISTS transit_occupancy.quarantine;
```

- reference:	Routes, stops, vehicles and trips
- bronze:	Raw Kafka events
- silver:	Clean, validated occupancy events
- gold:	    Business KPIs and aggregations
- quarantine:	Invalid or rejected events

```python
%sql

CREATE VOLUME IF NOT EXISTS transit_occupancy.reference.source_files;

CREATE VOLUME IF NOT EXISTS transit_occupancy.bronze.streaming_data;

CREATE VOLUME IF NOT EXISTS transit_occupancy.bronze.checkpoints;

CREATE VOLUME IF NOT EXISTS transit_occupancy.quarantine.rejected_data;
```
