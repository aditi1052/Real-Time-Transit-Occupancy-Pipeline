```python
import pandas as pd
from pathlib import Path
```

```python
Output_Path = Path("data/reference")
Output_Path.mkdir(parents=True, exist_ok=True)
```

```python
vehicles = []
routes = []
stops = []
route_stops = []
trips = []

# Generate vehicles
capacities = [40, 50, 60]

for vehicle_number in range(1, 21):
    vehicles.append(
        {
            "vehicle_id": f"BUS-{vehicle_number:03d}",
            "vehicle_type": "Standard",
            "vehicle_capacity": capacities[(vehicle_number - 1) % 3],
            "service_status": "ACTIVE"
        }
    )

pd.DataFrame(vehicles).to_csv(
    Output_Path / "vehicles.csv",
    index=False
)
```
